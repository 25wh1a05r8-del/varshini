```python
score = 87
if score >= 90:
    grade = 'A (Excellent)'
elif score >= 80:
    grade = 'B (Good)'
elif score >= 70:
    grade = 'C (Average)'
elif score >= 60:
    grade = 'D (Below Average)'
else:
    grade = 'F (Fail)'

print(f'Score: {score} → Grade: {grade}')
```

```python
n = 36
if n % 2 == 0:
    print(f'{n} is Even')
else:
    print(f'{n} is Odd')

if n % 3 == 0 and n % 5 == 0:
    print(f'{n} is divisible by both 3 and 5')
elif n % 3 == 0:
    print(f'{n} is divisible by 3 only')
elif n % 5 == 0:
    print(f'{n} is divisible by 5 only')
```

```python
n = 10
total = 0

for i in range(1, n + 1):
    total += i

formula_result = n * (n + 1) // 2

print(f'Sum (loop)    : {total}')
print(f'Sum (formula) : {formula_result}')
```

```python
n = 6
factorial = 1

for i in range(1, n + 1):
    factorial *= i
    print(f'   {i}!  = {factorial}')
```
