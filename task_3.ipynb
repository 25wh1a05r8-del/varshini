```python
print("\n--- Task 3: Core Library Usage ---")

import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")

# -----------------------------
# Global Data
# -----------------------------
DEVICE_TYPES = [
    'Temperature_Sensor',
    'Humidity_Sensor',
    'Pressure_Sensor',
    'Light_Sensor'
]

LOCATIONS = [
    'Lab1',
    'Lab2',
    'OfficeA',
    'OfficeB',
    'Warehouse'
]

READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}

# ==========================================================
# Task 1
# ==========================================================

print("\n--- Task 1: Data Generation and Basic Structures ---")

# -----------------------------
# Task 1.1
# -----------------------------

raw_sensor_data = []

start_time = datetime.datetime.now()

for i in range(100):

    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    minimum, maximum = READING_RANGES[device_type]

    value = round(random.uniform(minimum, maximum), 2)

    timestamp = (
        start_time + datetime.timedelta(minutes=i)
    ).strftime("%Y-%m-%d %H:%M:%S")

    device_id = f"DEV{i+1:03d}"

    raw_sensor_data.append(
        (
            timestamp,
            device_id,
            device_type,
            location,
            value
        )
    )

print("Generated", len(raw_sensor_data), "sensor readings.")

# -----------------------------
# Task 1.2
# -----------------------------

temperature_readings = []

for reading in raw_sensor_data:

    if reading[2] == "Temperature_Sensor":

        temperature_readings.append({

            "timestamp": reading[0],
            "device_id": reading[1],
            "location": reading[3],
            "temperature_celsius": reading[4]

        })

print("Temperature readings:", len(temperature_readings))

# -----------------------------
# Task 1.3
# -----------------------------

unique_locations = {reading[3] for reading in raw_sensor_data}

print("\nUnique Locations")
print(unique_locations)

# -----------------------------
# Task 1.4
# -----------------------------

device_stats = {}

for device in DEVICE_TYPES:
    device_stats[device] = (float("inf"), float("-inf"))

for reading in raw_sensor_data:

    device_type = reading[2]
    value = reading[4]

    minimum, maximum = device_stats[device_type]

    minimum = min(minimum, value)
    maximum = max(maximum, value)

    device_stats[device_type] = (minimum, maximum)

print("\nDevice Statistics")

for key, value in device_stats.items():
    print(key, ":", value)

# ==========================================================
# Task 1.5
# ==========================================================

print("\n--- Task 1.5: Advanced Indexing and Slicing ---")

sample_string = "PythonProgrammingIsFun"

# -----------------------------
# Task 1.5.1
# -----------------------------

sliced_readings_1 = raw_sensor_data[9:20]

sliced_readings_2 = raw_sensor_data[::5]

print("Readings from 10th to 20th:", len(sliced_readings_1))
print("Every 5th reading:", len(sliced_readings_2))

# -----------------------------
# Task 1.5.2
# -----------------------------

substring_1 = sample_string[6:17]

substring_2 = sample_string[:6][::-1]

print(substring_1)
print(substring_2)

# -----------------------------
# Task 1.5.3
# -----------------------------

fifth_reading_device_type = raw_sensor_data[4][2]

device_type_suffix = fifth_reading_device_type[-6:]

print(fifth_reading_device_type)
print(device_type_suffix)

# ==========================================================
# Task 2
# ==========================================================

print("\n--- Task 2: Function Definitions and Usage ---")

# -----------------------------
# Task 2.1
# -----------------------------

def get_readings_by_location(readings_list, target_location):

    result = []

    for reading in readings_list:

        if reading[3] == target_location:
            result.append(reading)

    return result

# -----------------------------
# Task 2.2
# -----------------------------

def calculate_average_temperature(temp_data_list):

    if len(temp_data_list) == 0:
        return 0.0

    total = 0

    for reading in temp_data_list:
        total += reading["temperature_celsius"]

    return total / len(temp_data_list)

# -----------------------------
# Task 2.3
# -----------------------------

def generate_report_line(reading_dict):

    return (
        f"[{reading_dict['timestamp']}] "
        f"Device {reading_dict['device_id']} "
        f"at {reading_dict['location']}: "
        f"{reading_dict['temperature_celsius']}°C"
    )

# -----------------------------
# Function Testing
# -----------------------------

lab1_readings = get_readings_by_location(raw_sensor_data, "Lab1")

print("\nNumber of readings from Lab1:", len(lab1_readings))

average_temperature = calculate_average_temperature(temperature_readings)

print("Average Temperature:", round(average_temperature, 2), "°C")

if temperature_readings:
    print("\nSample Report Line:")
    print(generate_report_line(temperature_readings[0]))
    # -----------------------------
# Task 3.1
# Create sensor_reports directory
# -----------------------------

reports_dir = "sensor_reports"

if not os.path.exists(reports_dir):
    os.makedirs(reports_dir)
    print(f"\nDirectory '{reports_dir}' created.")
else:
    print(f"\nDirectory '{reports_dir}' already exists.")

# -----------------------------
# Task 3.2
# Calculate duration between first and last timestamp
# -----------------------------

if len(raw_sensor_data) > 1:

    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]

    date_format = "%Y-%m-%d %H:%M:%S"

    first_dt = datetime.datetime.strptime(
        first_timestamp_str,
        date_format
    )

    last_dt = datetime.datetime.strptime(
        last_timestamp_str,
        date_format
    )

    duration = last_dt - first_dt

    days = duration.days
    total_seconds = int(duration.total_seconds())

    hours = (total_seconds // 3600) % 24
    minutes = (total_seconds % 3600) // 60

    print("\nDuration of Data Collection")
    print(f"Days    : {days}")
    print(f"Hours   : {hours}")
    print(f"Minutes : {minutes}")

else:
    print("Not enough data to calculate duration.")

# -----------------------------
# Task 3.3
# Generate 5 random integers
# -----------------------------

random_values = []

for i in range(5):
    random_values.append(random.randint(100, 200))

print("\nRandom Values")
print(random_values)
```
